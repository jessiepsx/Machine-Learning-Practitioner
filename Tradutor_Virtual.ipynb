```python
!pip install gTTS googletrans==3.1.0a0

from gtts import gTTS
from googletrans import Translator
from IPython.display import Audio, display
import time # Usado para dar uma pequena pausa no loop
```

```python
# ----------------------------------------------------
# DEFINIÇÃO DAS FUNÇÕES
# ----------------------------------------------------

def text_to_speech(text, lang):
    """Converte o texto traduzido em áudio e o reproduz."""
    try:
        tts = gTTS(text=text, lang=lang)
        tts.save("output.mp3")
        display(Audio("output.mp3", autoplay=True))
        print(f"-> Áudio reproduzido com sucesso na língua {lang}.")

    except Exception as e:
        print(f"Ocorreu um erro no Text-to-Speech: {e}")

def translate_text(text, target_lang):

    try:
        translator = Translator()
        # FORÇA a tradução de Português (src='pt') para o idioma de destino
        translated = translator.translate(text, src='pt', dest=target_lang)
        print(f"-> Traduzido para {target_lang}: {translated.text}")
        return translated.text

    except Exception as e:
        print(f"Ocorreu um erro durante a tradução: {e}")
        return text
```

```python
# ----------------------------------------------------
# O PROGRAMA PRINCIPAL
# ----------------------------------------------------

print("--- Tradutor Virtual ---")
print("Digite 'sair' a qualquer momento para encerrar o programa.\n")

while True:
    text = input("Digite o texto em PORTUGUÊS: ")

    if text.lower() == 'sair':
        print("\nPrograma encerrado. Até mais!")
        break

    TARGET_LANG = 'en'

    # 1. Traduz o texto
    translated_text = translate_text(text, target_lang=TARGET_LANG)

    # 2. Converte e reproduz o áudio
    text_to_speech(translated_text, lang=TARGET_LANG)

    # Pausa de 1 segundo para melhor experiência de uso
    time.sleep(1)
```
